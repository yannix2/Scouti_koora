**Goal:** Scrape real player stats (xG, xA, progressive passes, pressures, dribbles...) from FBref for all 32+ leagues.  
**Output:** `data/scraped/fbref_{stat_type}.csv` per stat type + `data/processed/fbref_master.csv`

---
⚠️ **Important before running:**
- This notebook makes ~190 HTTP requests (32 leagues × 6 stat types)
- A 4-second delay is built in between each request
- **Total runtime: ~15–25 minutes** — let it run, don't interrupt
- Results are auto-saved after each league so nothing is lost if it stops

```python
import sys, asyncio, pandas as pd
import time, random, nest_asyncio
from datetime import datetime
from pathlib import Path
from playwright.async_api import async_playwright

# ── Windows + Anaconda fix ────────────────────────────────
if sys.platform == 'win32':
    asyncio.set_event_loop_policy(asyncio.WindowsProactorEventLoopPolicy())
    loop = asyncio.new_event_loop()
    asyncio.set_event_loop(loop)
# ── Allow asyncio inside Jupyter ─────────────────────────
nest_asyncio.apply()

_playwright = None
_browser    = None
_context    = None
_page       = None

async def _start():
    global _playwright, _browser, _context, _page
    _playwright = await async_playwright().start()
    _browser    = await _playwright.chromium.launch(
        headless=True,
        args=['--no-sandbox', '--disable-dev-shm-usage']
    )
    _context = await _browser.new_context(
        viewport={'width': 1440, 'height': 900},
        user_agent='Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36',
        locale='en-US',
    )
    _page = await _context.new_page()

    # Block images/fonts to speed things up
    await _page.route(
        '**/*.{png,jpg,jpeg,gif,svg,woff,woff2,ttf,mp4,webm}',
        lambda r: asyncio.ensure_future(r.abort())
    )

    # Warm up session
    print('🌐 Warming up browser...')
    await _page.goto('https://fbref.com/en/', wait_until='domcontentloaded', timeout=30000)
    await asyncio.sleep(3)
    print('✅ Browser ready')


async def _stop():
    global _browser, _playwright
    if _browser:    await _browser.close()
    if _playwright: await _playwright.stop()
    print('🛑 Browser closed')


def start_browser():
    asyncio.get_event_loop().run_until_complete(_start())


def stop_browser():
    asyncio.get_event_loop().run_until_complete(_stop())


# ── Helpers ───────────────────────────────────────────────
def build_url(fbref_id, slug, season, stat_path):
    return (
        f'https://fbref.com/en/comps/{fbref_id}/{season}/{stat_path}/'
        f'players/{season}-{slug}-Stats'
    )


def flatten_columns(df):
    new_cols = []
    for top, bot in df.columns:
        top, bot = str(top).strip(), str(bot).strip()
        new_cols.append(bot if 'Unnamed' in top or top == '' else f'{top}_{bot}')
    df.columns = new_cols
    return df


def clean_table(df):
    for col in ['Player', 'Rk']:
        if col in df.columns:
            df = df[df[col] != col]
    return df.reset_index(drop=True)


# ── Core async scrape function ────────────────────────────
async def _scrape(fbref_id, slug, season, stat_type, stat_path,
                  comp_id, league_name, country, delay=5.0, retries=3):
    global _page
    url = build_url(fbref_id, slug, season, stat_path)

    for attempt in range(1, retries + 1):
        try:
            resp = await _page.goto(url, wait_until='domcontentloaded', timeout=35000)

            if resp.status == 429:
                wait = delay * (attempt * 4)
                print(f'      ⏳ Rate limited — waiting {wait:.0f}s...')
                await asyncio.sleep(wait)
                continue

            if resp.status == 403:
                print(f'      🔒 403 attempt {attempt} — waiting {delay*attempt:.0f}s...')
                await asyncio.sleep(delay * attempt)
                await _page.goto('https://fbref.com/en/', wait_until='domcontentloaded')
                await asyncio.sleep(3)
                continue

            if resp.status != 200:
                print(f'      ⚠️  HTTP {resp.status} — skipping')
                return pd.DataFrame()

            # Wait for the stats table
            try:
                await _page.wait_for_selector('table.stats_table', timeout=12000)
            except Exception:
                print(f'      ⚠️  No table found')
                return pd.DataFrame()

            html = await _page.content()

            try:
                tables = pd.read_html(html, header=[0, 1])
            except ValueError:
                return pd.DataFrame()

            if not tables:
                return pd.DataFrame()

            df = flatten_columns(tables[0].copy())
            df = clean_table(df)

            if 'Player' in df.columns:
                df = df[df['Player'].notna() & (df['Player'] != '')]

            df['fbref_season']    = season
            df['fbref_stat_type'] = stat_type
            df['kaggle_comp_id']  = comp_id
            df['league_name']     = league_name
            df['league_country']  = country
            df['scraped_at']      = datetime.now().strftime('%Y-%m-%d %H:%M')

            await asyncio.sleep(delay + random.uniform(0.5, 2.0))
            return df

        except Exception as e:
            print(f'      ❌ Error attempt {attempt}: {e}')
            await asyncio.sleep(delay * attempt)

    return pd.DataFrame()


# ── Public sync wrapper (called from notebook cells) ─────
def scrape_fbref_table(fbref_id, slug, season, stat_type, stat_path,
                       comp_id, league_name, country, delay=5.0, retries=3):
    return asyncio.get_event_loop().run_until_complete(
        _scrape(fbref_id, slug, season, stat_type, stat_path,
                comp_id, league_name, country, delay, retries)
    )


print('✅ Async Playwright scraper ready')
print('   Run start_browser() then proceed to cell 4')
```

## 🚀 3 — Checkpoint System
Saves progress after every league so the scraper can resume if interrupted.

```python
CHECKPOINT_FILE = SCRAPED_DIR / 'checkpoint.json'

def load_checkpoint() -> dict:
    """Load which (league, stat_type) pairs have already been scraped."""
    if CHECKPOINT_FILE.exists():
        with open(CHECKPOINT_FILE, 'r') as f:
            return json.load(f)
    return {'done': []}


def save_checkpoint(done_key: str, checkpoint: dict):
    """Mark a (league, stat_type) pair as done."""
    checkpoint['done'].append(done_key)
    with open(CHECKPOINT_FILE, 'w') as f:
        json.dump(checkpoint, f, indent=2)


def reset_checkpoint():
    """Call this to restart from scratch (deletes checkpoint)."""
    if CHECKPOINT_FILE.exists():
        CHECKPOINT_FILE.unlink()
    print('🗑️  Checkpoint cleared — will scrape everything from scratch')


checkpoint = load_checkpoint()
already_done = len(checkpoint['done'])
total_tasks  = len(LEAGUE_MAP) * len(STAT_TYPES)

print(f'📋 Checkpoint status:')
print(f'   Already scraped: {already_done} / {total_tasks} tasks')
print(f'   Remaining:       {total_tasks - already_done} tasks')

if already_done > 0:
    print(f'\n💡 Resuming from where we left off.')
    print(f'   To restart from scratch, run: reset_checkpoint()')
else:
    print(f'\n🆕 Fresh start — will scrape all {total_tasks} tasks')
```

## ⚡ 4 — Run the Full Scraper
> ☕ This takes 15–25 minutes. Each league is auto-saved to `data/scraped/` as it completes.

```python
checkpoint = load_checkpoint()
all_results = {}   # { stat_type: [df1, df2, ...] }

# Load any previously scraped files into memory
for stat_type in STAT_TYPES:
    cached = SCRAPED_DIR / f'fbref_{stat_type}_raw.csv'
    if cached.exists():
        all_results[stat_type] = [pd.read_csv(cached, low_memory=False)]
        print(f'  📂 Loaded cached: fbref_{stat_type}_raw.csv')
    else:
        all_results[stat_type] = []

print(f'\n🚀 Starting scrape for season {SEASON}...\n')

league_list = list(LEAGUE_MAP.items())

for comp_id, info in tqdm(league_list, desc='Leagues', unit='league'):
    league_name = info['name']
    country     = info['country']
    fbref_id    = info['fbref_id']
    slug        = info['slug']

    print(f'\n  🌍 [{comp_id}] {league_name} ({country})')

    for stat_type, stat_path in STAT_TYPES.items():
        done_key = f'{comp_id}_{stat_type}'

        # Skip if already done in a previous run
        if done_key in checkpoint['done']:
            print(f'     ✅ {stat_type:12s} — already done, skipping')
            continue

        df = scrape_fbref_table(
            fbref_id=fbref_id,
            slug=slug,
            season=SEASON,
            stat_type=stat_type,
            stat_path=stat_path,
            comp_id=comp_id,
            league_name=league_name,
            country=country,
            delay=4.5,
        )

        if df.empty:
            print(f'     ⚠️  {stat_type:12s} — no data returned')
        else:
            all_results[stat_type].append(df)
            print(f'     ✅ {stat_type:12s} — {len(df):,} players')

        save_checkpoint(done_key, checkpoint)

    # Save this stat type's accumulated data after every league
    for stat_type in STAT_TYPES:
        if all_results[stat_type]:
            combined = pd.concat(all_results[stat_type], ignore_index=True)
            out_path = SCRAPED_DIR / f'fbref_{stat_type}_raw.csv'
            combined.to_csv(out_path, index=False)

print('\n' + '='*55)
print('  ✅ SCRAPING COMPLETE')
print('='*55)

for stat_type in STAT_TYPES:
    if all_results[stat_type]:
        total_rows = sum(len(d) for d in all_results[stat_type])
        path = SCRAPED_DIR / f'fbref_{stat_type}_raw.csv'
        print(f'  💾 fbref_{stat_type}_raw.csv → {total_rows:,} rows')
```

## 🧹 6 — Clean Each Stat Table

```python
def clean_fbref_df(df: pd.DataFrame, stat_type: str) -> pd.DataFrame:
    """
    Cleans a raw FBref table:
    - Standardizes key column names across all stat types
    - Converts numeric columns
    - Removes duplicate players within same league
    - Adds normalized player name for merging
    """
    import unicodedata, re

    df = df.copy()

    # ── 1. Standardize core identity columns ──────────────
    rename_map = {}

    # Player name — may be 'Player' in all stat types
    for col in df.columns:
        low = col.lower()
        if col == 'Player':              rename_map[col] = 'player_name'
        elif col == 'Squad':             rename_map[col] = 'squad'
        elif col == 'Nation':            rename_map[col] = 'nation'
        elif col == 'Pos':               rename_map[col] = 'pos'
        elif col == 'Age':               rename_map[col] = 'age_str'
        elif col == 'Born':              rename_map[col] = 'birth_year'

    df.rename(columns=rename_map, inplace=True)

    # ── 2. Convert all numeric columns ────────────────────
    identity_cols = {'player_name','squad','nation','pos','age_str',
                     'birth_year','fbref_season','fbref_stat_type',
                     'kaggle_comp_id','league_name','league_country','scraped_at'}

    for col in df.columns:
        if col not in identity_cols:
            df[col] = pd.to_numeric(df[col], errors='coerce')

    # ── 3. Parse age (FBref format: "23-145" = 23 years 145 days) ─
    if 'age_str' in df.columns:
        df['age'] = df['age_str'].str.split('-').str[0].apply(
            pd.to_numeric, errors='coerce'
        )

    # ── 4. Prefix stat columns to avoid merge conflicts ───
    stat_cols = [c for c in df.columns if c not in identity_cols | {'age'}]
    prefix_map = {}
    for col in stat_cols:
        if not col.startswith(stat_type + '_'):
            prefix_map[col] = f'{stat_type}_{col}'
    df.rename(columns=prefix_map, inplace=True)

    # ── 5. Normalize player name for joining ─────────────
    def normalize(name):
        if pd.isna(name): return ''
        name = unicodedata.normalize('NFD', str(name))
        name = ''.join(c for c in name if unicodedata.category(c) != 'Mn')
        name = re.sub(r'[^a-zA-Z\s]', '', name).lower().strip()
        return re.sub(r'\s+', ' ', name)

    df['name_norm'] = df['player_name'].apply(normalize)

    # ── 6. Drop rows with no player name ─────────────────
    df = df[df['name_norm'] != ''].reset_index(drop=True)

    # ── 7. Deduplicate same player in same league ─────────
    # Keep row with most minutes played
    min_col = next((c for c in df.columns if 'min' in c.lower() and stat_type in c.lower()), None)
    if min_col:
        df = (
            df.sort_values(min_col, ascending=False)
              .drop_duplicates(subset=['name_norm','league_name'], keep='first')
              .reset_index(drop=True)
        )

    return df


# Apply cleaning to all scraped tables
cleaned = {}
for stat_type, df in scraped.items():
    cleaned[stat_type] = clean_fbref_df(df, stat_type)
    out = SCRAPED_DIR / f'fbref_{stat_type}_clean.csv'
    cleaned[stat_type].to_csv(out, index=False)
    print(f'✅ {stat_type:12s} → {len(cleaned[stat_type]):,} rows after cleaning  (saved)')
```

## 🔗 7 — Merge All Stat Types → FBref Master

```python
# Keys used to join stat tables
JOIN_KEYS = ['name_norm', 'squad', 'league_name', 'league_country', 'kaggle_comp_id', 'fbref_season']

if 'standard' not in cleaned or cleaned['standard'].empty:
    raise ValueError('Standard stats not found — run the scraper first (cell 4)')

# Start from standard stats as the base
fbref_master = cleaned['standard'].copy()
print(f'Base (standard): {fbref_master.shape[0]:,} rows')

# Merge each additional stat type
for stat_type in ['shooting', 'passing', 'defense', 'possession', 'misc']:
    if stat_type not in cleaned or cleaned[stat_type].empty:
        print(f'  ⚠️  {stat_type} — skipping (empty)')
        continue

    # Keep only stat columns + join keys from each table
    stat_cols    = [c for c in cleaned[stat_type].columns if c.startswith(stat_type + '_')]
    available_keys = [k for k in JOIN_KEYS if k in cleaned[stat_type].columns]

    merge_df = cleaned[stat_type][available_keys + stat_cols]

    fbref_master = fbref_master.merge(
        merge_df,
        on=available_keys,
        how='left'
    )
    print(f'  + {stat_type:12s}: now {fbref_master.shape[1]} columns')

print(f'\n✅ FBref master shape: {fbref_master.shape[0]:,} players × {fbref_master.shape[1]} columns')
fbref_master.to_csv(SCRAPED_DIR / 'fbref_master.csv', index=False)
print(f'💾 Saved → data/scraped/fbref_master.csv')
```

## 🔗 8 — Merge FBref + Transfermarkt → Final Scouting Dataset

```python
import unicodedata, re

def normalize_name(name):
    if pd.isna(name): return ''
    name = unicodedata.normalize('NFD', str(name))
    name = ''.join(c for c in name if unicodedata.category(c) != 'Mn')
    name = re.sub(r'[^a-zA-Z\s]', '', name).lower().strip()
    return re.sub(r'\s+', ' ', name)

# ── Load Transfermarkt master from notebook 01 ───────────
players  = pd.read_csv('../data/raw/players.csv',       low_memory=False)
apps_raw = pd.read_csv('../data/raw/appearances.csv',   low_memory=False)
vals     = pd.read_csv('../data/raw/player_valuations.csv', low_memory=False)
transfers= pd.read_csv('../data/raw/transfers.csv',     low_memory=False)

# Re-compute career stats
career = apps_raw.groupby('player_id').agg(
    career_games    = ('appearance_id', 'count'),
    career_goals    = ('goals',          'sum'),
    career_assists  = ('assists',         'sum'),
    career_minutes  = ('minutes_played',  'sum'),
    career_yellows  = ('yellow_cards',    'sum'),
    career_reds     = ('red_cards',       'sum'),
).reset_index()
career['career_goals_p90']   = (career['career_goals']   / (career['career_minutes'] / 90)).round(3)
career['career_assists_p90'] = (career['career_assists']  / (career['career_minutes'] / 90)).round(3)

# Last 12 months form
apps_raw['date'] = pd.to_datetime(apps_raw['date'], errors='coerce')
cutoff = apps_raw['date'].max() - pd.DateOffset(months=12)
recent = apps_raw[apps_raw['date'] >= cutoff].groupby('player_id').agg(
    form_games   = ('appearance_id', 'count'),
    form_goals   = ('goals',          'sum'),
    form_assists = ('assists',         'sum'),
    form_minutes = ('minutes_played',  'sum'),
).reset_index()

# Latest market valuation
vals['date'] = pd.to_datetime(vals['date'], errors='coerce')
latest_val = (
    vals.sort_values('date').groupby('player_id').last().reset_index()
    [['player_id','market_value_in_eur','date']]
    .rename(columns={'market_value_in_eur': 'latest_value_eur',
                     'date': 'valuation_date'})
)

# Transfer count + max fee
transfer_stats = transfers.groupby('player_id').agg(
    total_transfers = ('transfer_fee', 'count'),
    max_fee_eur     = ('transfer_fee', 'max'),
).reset_index()

# Build TM master
players['date_of_birth'] = pd.to_datetime(players['date_of_birth'], errors='coerce')
players['age'] = ((pd.Timestamp.now() - players['date_of_birth']).dt.days / 365.25).round(1)
players['name_norm'] = players['name'].apply(normalize_name)

tm = (
    players
    .merge(career,       on='player_id', how='left')
    .merge(recent,       on='player_id', how='left')
    .merge(latest_val,   on='player_id', how='left')
    .merge(transfer_stats,on='player_id', how='left')
)

print(f'TM master: {tm.shape[0]:,} players × {tm.shape[1]} columns')

# ── Merge FBref + TM on normalized name ─────────────────
scouting = fbref_master.merge(
    tm,
    on='name_norm',
    how='left',
    suffixes=('_fbref', '_tm')
)

matched = scouting['player_id'].notna().sum()
total   = len(scouting)
print(f'\n📊 Merge results:')
print(f'   Total FBref players:  {total:,}')
print(f'   Matched with TM data: {matched:,} ({matched/total*100:.1f}%)')
print(f'   Unmatched:            {total-matched:,} ({(total-matched)/total*100:.1f}%)')
```

## ⚙️ 9 — Feature Engineering (per-90 KPIs + composite scores)

```python
df = scouting.copy()

# ── Helper: safe per-90 from a column ────────────────────
def per90(col):
    if col not in df.columns: return pd.Series(np.nan, index=df.index)
    minutes = df.get('standard_Playing Time_Min', df.get('career_minutes', pd.Series(np.nan, index=df.index)))
    p90 = minutes.replace(0, np.nan) / 90
    return (pd.to_numeric(df[col], errors='coerce') / p90).round(3)

# ── Minutes played (standardized name) ───────────────────
min_col = next((c for c in df.columns if 'Min' in c and 'standard' in c), None)
if min_col:
    df['minutes_played'] = pd.to_numeric(df[min_col], errors='coerce')
else:
    df['minutes_played'] = df.get('career_minutes', np.nan)

df['p90'] = (df['minutes_played'].replace(0, np.nan) / 90)

# ── Attacking KPIs ────────────────────────────────────────
gls_col = next((c for c in df.columns if c.endswith('_Gls') and 'standard' in c), None)
ast_col = next((c for c in df.columns if c.endswith('_Ast') and 'standard' in c), None)
xg_col  = next((c for c in df.columns if 'xG' in c      and 'standard' in c), None)
xa_col  = next((c for c in df.columns if 'xAG' in c     and 'standard' in c), None)
sot_col = next((c for c in df.columns if 'SoT%' in c    and 'shooting' in c), None)

if gls_col: df['goals_p90']   = (pd.to_numeric(df[gls_col], errors='coerce') / df['p90']).round(3)
if ast_col: df['assists_p90'] = (pd.to_numeric(df[ast_col], errors='coerce') / df['p90']).round(3)
if xg_col:  df['xG_p90']      = (pd.to_numeric(df[xg_col],  errors='coerce') / df['p90']).round(3)
if xa_col:  df['xA_p90']      = (pd.to_numeric(df[xa_col],  errors='coerce') / df['p90']).round(3)

# ── Defensive KPIs ────────────────────────────────────────
tkl_col  = next((c for c in df.columns if 'Tkl' in c    and 'defense' in c and 'Won' not in c), None)
int_col  = next((c for c in df.columns if '_Int' in c   and 'defense' in c), None)
prs_col  = next((c for c in df.columns if 'Press' in c  and 'defense' in c), None)
blk_col  = next((c for c in df.columns if 'Blocks' in c and 'defense' in c), None)

if tkl_col: df['tackles_p90']      = (pd.to_numeric(df[tkl_col],  errors='coerce') / df['p90']).round(3)
if int_col: df['interceptions_p90']= (pd.to_numeric(df[int_col],  errors='coerce') / df['p90']).round(3)
if prs_col: df['pressures_p90']    = (pd.to_numeric(df[prs_col],  errors='coerce') / df['p90']).round(3)
if blk_col: df['blocks_p90']       = (pd.to_numeric(df[blk_col],  errors='coerce') / df['p90']).round(3)

# ── Progression KPIs ──────────────────────────────────────
prog_carry = next((c for c in df.columns if 'PrgC' in c and 'possession' in c), None)
prog_pass  = next((c for c in df.columns if 'PrgP' in c and 'passing'    in c), None)
kp_col     = next((c for c in df.columns if 'KP'   in c and 'passing'    in c), None)

if prog_carry: df['prog_carries_p90'] = (pd.to_numeric(df[prog_carry], errors='coerce') / df['p90']).round(3)
if prog_pass:  df['prog_passes_p90']  = (pd.to_numeric(df[prog_pass],  errors='coerce') / df['p90']).round(3)
if kp_col:     df['key_passes_p90']   = (pd.to_numeric(df[kp_col],     errors='coerce') / df['p90']).round(3)

# ── Composite Scores (0–10 scale normalized) ─────────────
def minmax(series):
    """Normalize a series to 0–10."""
    mn, mx = series.min(), series.max()
    if mx == mn: return pd.Series(5.0, index=series.index)
    return ((series - mn) / (mx - mn) * 10).round(2)

attack_parts  = [df.get(c) for c in ['goals_p90','xG_p90','assists_p90','xA_p90'] if c in df.columns]
defense_parts = [df.get(c) for c in ['tackles_p90','interceptions_p90','pressures_p90'] if c in df.columns]
progress_parts= [df.get(c) for c in ['prog_carries_p90','prog_passes_p90','key_passes_p90'] if c in df.columns]

if attack_parts:
    df['attack_score']   = minmax(pd.concat(attack_parts,   axis=1).mean(axis=1))
if defense_parts:
    df['defense_score']  = minmax(pd.concat(defense_parts,  axis=1).mean(axis=1))
if progress_parts:
    df['progress_score'] = minmax(pd.concat(progress_parts, axis=1).mean(axis=1))

# ── Value Efficiency ──────────────────────────────────────
val_col = 'latest_value_eur' if 'latest_value_eur' in df.columns else 'market_value_in_eur'
if val_col in df.columns and gls_col and ast_col:
    df['value_efficiency'] = (
        (pd.to_numeric(df[gls_col], errors='coerce').fillna(0) +
         pd.to_numeric(df[ast_col], errors='coerce').fillna(0)) /
        (pd.to_numeric(df[val_col], errors='coerce').replace(0, np.nan) / 1_000_000)
    ).round(3)

kpi_cols = [c for c in df.columns if any(x in c for x in ['_p90','_score','efficiency'])]
print(f'✅ KPIs computed: {len(kpi_cols)} new columns')
for c in kpi_cols:
    non_null = df[c].notna().sum()
    print(f'   {c:<30} {non_null:,} non-null values')
```

## 🧹 10 — Final Clean & Save Master Dataset

```python
# ── Keep only players with enough playing time ────────────
df_final = df[df['minutes_played'].fillna(0) >= 90].copy()
print(f'Players with 90+ minutes: {len(df_final):,} (removed {len(df)-len(df_final):,} bench players)')

# ── Drop noisy / redundant meta columns ───────────────────
drop_patterns = [
    'Unnamed', 'Rk', 'Matches', 'url', 'image', 'player_code',
    'city_of_birth', 'scraped_at', 'fbref_stat_type',
]
drop_cols = [
    c for c in df_final.columns
    if any(p.lower() in c.lower() for p in drop_patterns)
]
df_final.drop(columns=drop_cols, inplace=True, errors='ignore')

# ── Deduplicate ───────────────────────────────────────────
dedup_key = ['name_norm', 'league_name']
before = len(df_final)
df_final = df_final.drop_duplicates(subset=dedup_key, keep='first').reset_index(drop=True)
print(f'Deduplication: {before:,} → {len(df_final):,} rows')

# ── Sort by value descending ──────────────────────────────
val_col = 'latest_value_eur' if 'latest_value_eur' in df_final.columns else 'market_value_in_eur'
if val_col in df_final.columns:
    df_final.sort_values(val_col, ascending=False, inplace=True)

# ── Save ──────────────────────────────────────────────────
out_path = PROCESSED_DIR / 'master_scouting_dataset.csv'
df_final.to_csv(out_path, index=False)

print(f'\n{"="*55}')
print(f'  ✅ MASTER SCOUTING DATASET READY')
print(f'{"="*55}')
print(f'  Players:  {len(df_final):,}')
print(f'  Columns:  {len(df_final.columns)}')
print(f'  Leagues:  {df_final["league_name"].nunique()}')
print(f'  File:     data/processed/master_scouting_dataset.csv')

# Preview
preview_cols = ['player_name','age','pos','squad','league_name',
                val_col,'goals_p90','assists_p90','attack_score','defense_score']
preview_cols = [c for c in preview_cols if c in df_final.columns]
print(f'\n📊 Sample rows:')
print(df_final[preview_cols].head(10).to_string(index=False))
```
